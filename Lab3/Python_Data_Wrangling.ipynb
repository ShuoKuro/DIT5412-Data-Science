```python
import pandas as pd
```

## 1. Merging Data

`pd.merge()` joins two DataFrames on a shared column (`subject_id` here), similar to VLOOKUP in Excel or JOIN in databases.

First, create and print the two original tables:

```python
left = pd.DataFrame({
         'id':[1,2,3,4,5],
         'Name': ['Alex', 'Amy', 'Allen', 'Alice', 'Ayoung'],
         'subject_id':['sub1','sub2','sub4','sub6','sub5']})
right = pd.DataFrame({
         'id':[1,2,3,4,5],
         'Name': ['Billy', 'Brian', 'Bran', 'Bryce', 'Betty'],
         'subject_id':['sub2','sub4','sub3','sub6','sub5']})
print(left)
print(right)
```

Merge on `subject_id` (inner join: only keys present in BOTH tables are kept):

```python
print(pd.merge(left, right, on='subject_id'))
```

**Observation**: only 4 rows remain after merging. `sub1` and `sub3` are dropped because each appears in only one of the two tables.

## 2. Grouping Data

`df.groupby('Year')` splits the data into groups by year, and `get_group()` retrieves one of them.

```python
ipl_data = {'Team': ['Riders','Riders','Devils','Devils','Kings','kings',
                     'Kings','Kings','Riders','Royals','Royals','Riders'],
            'Rank': [1,2,2,3,3,4,1,1,2,4,1,2],
            'Year': [2014,2015,2014,2015,2014,2015,2016,2017,2016,2014,2015,2017],
            'Points':[876,789,863,673,741,812,756,788,694,701,804,690]}
df = pd.DataFrame(ipl_data)
df
```

```python
grouped = df.groupby('Year')
grouped.get_group(2014)   # retrieve the 2014 group
```

**Observation**: the 2014 group contains 4 rows (Riders, Devils, Kings, Royals).

## 3. Concatenating Data

`pd.concat()` stacks tables with the same structure along an axis (it does NOT renumber the index).

```python
one = pd.DataFrame({
        'Name': ['Alex','Amy','Allen','Alice','Ayoung'],
        'subject_id':['sub1','sub2','sub4','sub6','sub5'],
        'Marks_scored':[98,90,87,69,78]}, index=[1,2,3,4,5])
two = pd.DataFrame({
        'Name': ['Billy','Brian','Bran','Bryce','Betty'],
        'subject_id':['sub2','sub4','sub3','sub6','sub5'],
        'Marks_scored':[89,80,79,97,88]}, index=[1,2,3,4,5])
print(one)
print(two)
```

```python
print(pd.concat([one, two]))
```

**Observation**: the result has 10 rows, with index 1-5 appearing twice, showing that `concat` simply stacks the tables without handling duplicates.

## Summary

| Operation | Function | Purpose |
|---|---|---|
| Merging | `pd.merge()` | Join two tables horizontally on a shared key |
| Grouping | `df.groupby()` | Split data into groups, then operate on each group |
| Concatenating | `pd.concat()` | Stack tables along an axis |
