## 2. Computation on NumPy Arrays (Part 2 of 5)

```python
import numpy as np

# Create 2 same size numeric arrays
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

# Add the 2 arrays
add_result = a + b

# Print the result
print('Addition:', add_result)

# Subtract one array from the other
sub_result = a - b

# Print the result
print('Subtraction:', sub_result)
```

## 3. Computation on NumPy Arrays (Part 3 of 5)

```python
import numpy as np

# Create a numeric array
arr = np.array([1, 2, 3, 4])

# Use the prod function
x = np.prod(arr)

# Print the result
print('Product:', x)

# Use the diff function
d = np.diff(arr)

# Print the array
print('Difference:', d)
```

## 5. Computation on NumPy Arrays (Part 5 of 5)

```python
from numpy import random
import numpy as np

# Create a random number from 0 to 1000
x = random.randint(1000)

# Print the result
print('Random number from 0 to 1000:', x)

arr = np.array([5, 2, 1, 3, 0])

# Shuffle the array
random.shuffle(arr)

# Print the result
print('After first shuffle:', arr)

# Shuffle the array again
random.shuffle(arr)

# Print the result
print('After second shuffle:', arr)
```

## 6. Indexing & Sorting in NumPy

```python
from numpy import random
import numpy as np

arr = np.array([5, 2, 1, 3, 0])

# Sort the array
sorted_arr = np.sort(arr)

# Print the result
print('Sorted array:', sorted_arr)

# Slice the last two elements
last_two = sorted_arr[-2:]

# Print the result
print('Last two elements:', last_two)
```

## 7. Probability Distribution in NumPy

```python
from numpy import random
import matplotlib.pyplot as plt

# Create a binomial distribution plot with p = 0.3
x = random.binomial(n=10, p=0.3, size=1000)

plt.hist(x, bins=range(0, 12), edgecolor='black', align='left', rwidth=0.8)
plt.title('Binomial Distribution (n=10, p=0.3, size=1000)')
plt.xlabel('Number of successes')
plt.ylabel('Frequency')

# Print the graphic result
plt.show()
```

## 8. Structured Data using NumPy

Following the commands on [Python Data Science Handbook — 02.09 Structured Data: NumPy's Structured Arrays](https://jakevdp.github.io/PythonDataScienceHandbook/02.09-structured-data-numpy.html), from Step [1] to Step [13].

```python
# Step [1]
import numpy as np
```

```python
# Step [2]
name = ['Alice', 'Bob', 'Cathy', 'Doug']
age = [25, 45, 37, 19]
weight = [55.0, 85.5, 68.0, 61.5]
```

```python
# Step [3]
x = np.zeros(4, dtype=int)
```

```python
# Step [4]
# Use a compound data type for structured arrays
data = np.zeros(4, dtype={'names': ('name', 'age', 'weight'),
                          'formats': ('U10', 'i4', 'f8')})
print(data.dtype)
```

```python
# Step [5]
data['name'] = name
data['age'] = age
data['weight'] = weight
print(data)
```

```python
# Step [6]
# Get all names
print(data['name'])
```

```python
# Step [7]
# Get first row of data
print(data[0])
```

```python
# Step [8]
# Get the name from the last row
print(data[-1]['name'])
```

```python
# Step [9]
# Get names where age is under 30
print(data[data['age'] < 30]['name'])
```

```python
# Step [10]
print(np.dtype({'names': ('name', 'age', 'weight'),
                'formats': ('U10', 'i4', 'f8')}))
```

```python
# Step [11]
print(np.dtype({'names': ('name', 'age', 'weight'),
                'formats': ((np.str_, 10), int, np.float32)}))
```

```python
# Step [12]
print(np.dtype([('name', 'S10'), ('age', 'i4'), ('weight', 'f8')]))
```

```python
# Step [13]
print(np.dtype('S10,i4,f8'))
```
